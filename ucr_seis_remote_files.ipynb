# Setting up pelicanfs, the python API

```python
import fsspec
from pelicanfs.core import OSDFFileSystem, PelicanMap
```

# Inspect the contents of ucr_seis

```python
fs = OSDFFileSystem()
dirpath = "/ndp/public/ucr_seis"

for obj in fs.ls(dirpath):

    print(obj["name"])

    if obj["type"] == "directory":

        for subobj in fs.ls(obj["name"]):

            print(subobj["name"])
```

# Load the torch state dict from a remote file

```python
import torch
```

```python
with fsspec.open("osdf:///ndp/public/ucr_seis/models/8700_17-319.ckpt") as fi:

    print("The remote file:")
    print(fi)

    state = torch.load(fi, map_location=torch.device("cpu"))

    print("The keys of the state dict:")
    for key in state.keys():
        print(" ", key)
```
